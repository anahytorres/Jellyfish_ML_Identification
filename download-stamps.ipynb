```python
from tqdm import tqdm
import urllib.request
import pandas as pd
import numpy as np
import os
import sys

df = pd.read_csv('/content/drive/MyDrive/Tesis/final_classified_sample_clusters.csv') # where final classified sample clusters is
```

```python
name = df.iloc[:, 0].values
RA_coord = df['ra'].astype(float).values
Dec_coord = df['dec'].astype(float).values

print(name)

scale = 0.25


for i in tqdm(np.arange(len(RA_coord))):
    outfile = f'/content/drive/MyDrive/Tesis/try/{name[i]}_{scale:.2f}_stamp.jpg' # change output directory
    if not os.path.exists(outfile):
        vurl = (
            f"https://www.legacysurvey.org/viewer/cutout.jpg"
            f"?ra={RA_coord[i]:f}&dec={Dec_coord[i]:f}"
            f"&layer=ls-dr10"
            f"&pixscale={scale:.2f}"
            f"&bands=griz"
        )
        urllib.request.urlretrieve(vurl, outfile)
```
